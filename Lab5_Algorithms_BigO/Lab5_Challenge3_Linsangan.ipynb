```python
import time
import random

LAST_NAME = "Linsangan"
STUDENT_ID = "TPUM_26_1846"
FAVORITE_ARTIST = " Steve Perry"

def bubble_sort(arr):
    """Sorts an array using Bubble Sort, tracking comparisons and movements."""
    temp_arr = arr.copy()
    n = len(temp_arr)
    comparisons = 0
    movements = 0  
    
    for i in range(n):
        for j in range(0, n - i - 1):
            comparisons += 1
            if temp_arr[j] > temp_arr[j + 1]:
                temp_arr[j], temp_arr[j + 1] = temp_arr[j + 1], temp_arr[j]
                movements += 3 
    return temp_arr, comparisons, movements


def selection_sort(arr):
    """Sorts an array using Selection Sort, tracking comparisons and movements."""
    temp_arr = arr.copy()
    n = len(temp_arr)
    comparisons = 0
    movements = 0
    
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if temp_arr[j] < temp_arr[min_idx]:
                min_idx = j
        if min_idx != i:
            temp_arr[i], temp_arr[min_idx] = temp_arr[min_idx], temp_arr[i]
            movements += 3
    return temp_arr, comparisons, movements


def insertion_sort(arr):
    """Sorts an array using Insertion Sort, tracking comparisons and movements."""
    temp_arr = arr.copy()
    n = len(temp_arr)
    comparisons = 0
    movements = 0
    
    for i in range(1, n):
        key = temp_arr[i]
        movements += 1 
        j = i - 1
        
        while j >= 0:
            comparisons += 1
            if temp_arr[j] > key:
                temp_arr[j + 1] = temp_arr[j]
                movements += 1 
                j -= 1
            else:
                break
        temp_arr[j + 1] = key
        movements += 1  # Placing key back
    return temp_arr, comparisons, movements

if __name__ == "__main__":
    seed_val = sum(int(d) for d in STUDENT_ID if d.isdigit()) + len(LAST_NAME)
    random.seed(seed_val)
    
    base_orders = random.sample(range(100, 999), 10)
    
    sorted_input = sorted(base_orders)
    reverse_input = sorted(base_orders, reverse=True)
    random_input = base_orders.copy()
    random.shuffle(random_input)
    
    inputs_dict = {
        "Sorted": sorted_input,
        "Reverse": reverse_input,
        "Random": random_input
    }
    
    algorithms = {
        "Bubble Sort": bubble_sort,
        "Selection Sort": selection_sort,
        "Insertion Sort": insertion_sort
    }
    
    print("Data:")
    print(f"Generated Base Orders (Random):  {random_input}")
    print(f"Generated Sorted Configuration: {sorted_input}")
    print(f"Generated Reverse Configuration:{reverse_input}\n")
    
    for algo_name, algo_func in algorithms.items():
        print(f"--- {algo_name} ---")
        for condition_name, current_data in inputs_dict.items():
            
            start_time = time.perf_counter()
            sorted_res, comps, moves = algo_func(current_data)
            end_time = time.perf_counter()
            exec_time_ms = (end_time - start_time) * 1000
            
            is_valid = sorted_res == sorted(base_orders)
            verification = "Passed" if is_valid else "Failed"
            
            print(f"[{condition_name} Input]")
            print(f"  Comparisons: {comps}")
            print(f"  Movements:   {moves}")
            print(f"  Exec Time:   {exec_time_ms:.4f} ms")
            print(f"  Verification: Sorting {verification}")
        print()
```
