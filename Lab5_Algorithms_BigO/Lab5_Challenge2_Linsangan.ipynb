```python
LAST_NAME = "Linsangan"       
STUDENT_ID = "TUPM-26-1846"      
FAVORITE_ARTIST = "Steve Perry" 

NAME_LENGTH = len(LAST_NAME)

ARTIST_LENGTH = len(FAVORITE_ARTIST)

SEED_DIGIT = int(STUDENT_ID[-1])

ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
def merge_enrollment_datasets(dept_a, dept_b):
    """
    Consolidates two sorted datasets into one sorted collection
    using a linear two-pointer merge strategy, tracking the number of comparisons.
    """
    consolidated = []
    i = 0  
    j = 0  
    comparison_count = 0
  
    while i < len(dept_a) and j < len(dept_b):
        comparison_count += 1
        if dept_a[i] <= dept_b[j]:
            consolidated.append(dept_a[i])
            i += 1
        else:
            consolidated.append(dept_b[j])
            j += 1
            
    while i < len(dept_a):
        consolidated.append(dept_a[i])
        i += 1
        
    while j < len(dept_b):
        consolidated.append(dept_b[j])
        j += 1
        
    return consolidated, comparison_count

def verify_ordering(dataset):
    """Checks if the array is strictly sorted in ascending order."""
    return all(dataset[k] <= dataset[k + 1] for k in range(len(dataset) - 1))


if __name__ == "__main__":
    print("Input:")
    print(f"NAME_LENGTH   = {NAME_LENGTH}")
    print(f"ARTIST_LENGTH = {ARTIST_LENGTH}")
    print(f"SEED_DIGIT    = {SEED_DIGIT}")
    print(f"ID_SUM        = {ID_SUM}\n")

    base_a = SEED_DIGIT + ID_SUM
    base_b = ARTIST_LENGTH + NAME_LENGTH

    department_a = [base_a, base_a + 4, base_a + 6, base_a + 8, base_a + 11]
    department_b = [base_b, base_b + 2, base_b + 10, base_b + 12, base_b + 22]
    
    records_before = len(department_a) + len(department_b)
    
    print("Data:")
    print(f"Department A Dataset: {department_a}")
    print(f"Department B Dataset: {department_b}")
    
    consolidated_list, total_comparisons = merge_enrollment_datasets(department_a, department_b)
    records_after = len(consolidated_list)
    
    print(f"Consolidated Dataset: {consolidated_list}")
    print(f"Number of Records Before Consolidation: {records_before}")
    print(f"Number of Records After Consolidation:  {records_after}")
    print(f"Comparison Count: {total_comparisons}")
    
    is_ordered = verify_ordering(consolidated_list)
    if is_ordered and (records_before == records_after):
        print("Ordering Verification: Verified")
    else:
        print("Ordering Verification: Failed")
```
