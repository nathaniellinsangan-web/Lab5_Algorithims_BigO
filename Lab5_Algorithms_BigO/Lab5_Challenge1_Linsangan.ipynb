```python
def calculate_student_inputs(last_name, student_id, favorite_artist):
    last_name = last_name.upper().replace(" ", "")
    favorite_artist = favorite_artist.upper().replace(" ", "")

    digits = [int(d) for d in str(student_id) if d.isdigit()]
    if not digits:
        raise ValueError("Student ID must contain numbers.")
        
    seed_digit = digits[-1]
    id_sum = sum(digits)

    name_length = len(last_name)
    artist_length = len(favorite_artist)
    
    artist_sum = sum(ord(char) - ord('A') + 1 for char in favorite_artist if char.isalpha())
    
    return {
        "SEED_DIGIT": seed_digit,
        "ID_SUM": id_sum,
        "NAME_LENGTH": name_length,
        "ARTIST_LENGTH": artist_length,
        "ARTIST_SUM": artist_sum
    }

def generate_datasets(rules):
    base_start = rules["ID_SUM"] + rules["NAME_LENGTH"]
    size = 15 + (rules["ARTIST_LENGTH"] % 10)  
    step = (rules["SEED_DIGIT"] % 4) + 2       
    
    original_dataset = [base_start + (i * step) for i in range(size)]

    pivot = rules["SEED_DIGIT"] if rules["SEED_DIGIT"] > 0 else len(original_dataset) // 3
    pivot = pivot % len(original_dataset)
    
    rotated_dataset = original_dataset[pivot:] + original_dataset[:pivot]
    
    target_incident = original_dataset[(rules["ARTIST_SUM"] + rules["ID_SUM"]) % size]
    
    return original_dataset, rotated_dataset, target_incident

def linear_search(arr, target):
    comparisons = 0
    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            return i, comparisons
    return -1, comparisons

def modified_binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    comparisons = 0
    
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        
        if arr[mid] == target:
            return mid, comparisons
            
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
                
    return -1, comparisons

LAST_NAME = "Linsangan"
STUDENT_ID = "TUPM-26-1846"
FAVORITE_ARTIST = "Steve Perry"

print("Student Input")
rules = calculate_student_inputs(LAST_NAME, STUDENT_ID, FAVORITE_ARTIST)
for key, val in rules.items():
    print(f"{key}: {val}")

orig_arr, rot_arr, target = generate_datasets(rules)

print("\nDatasets:")
print(f"Original Dataset:  {orig_arr}")
print(f"Rotated Dataset:   {rot_arr}")
print(f"Target Incident:   {target}")

bs_pos, bs_comps = modified_binary_search(rot_arr, target)
ls_pos, ls_comps = linear_search(rot_arr, target)

print("\nAnalysis:")
print(f"Target Incident to find: {target}")
print(f"Modified Binary Search Position: {bs_pos} (Found in {bs_comps} comparisons)")
print(f"Linear Search Position:          {ls_pos} (Found in {ls_comps} comparisons)")

print("\nComplexity:")
print("Modified Binary Search:")
print("  - Time Complexity:  O(log n) -> Sublinear / Logarithmic time")
print("  - Space Complexity: O(1)     -> Constant space (Iterative approach)")
print("Linear Search:")
print("  - Time Complexity:  O(n)   -> Linear time")
print("  - Space Complexity: O(1)     -> Constant space")
```
